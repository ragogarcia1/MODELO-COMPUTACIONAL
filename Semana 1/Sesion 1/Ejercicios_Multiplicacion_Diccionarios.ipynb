# Ejercicios de Multiplicación de Diccionarios en Python

```python
# 1) Multiplicación valor a valor entre dos diccionarios
d1 = {"a": 2, "b": 3, "c": 4}
d2 = {"a": 5, "b": 6, "c": 7}

resultado = {}
for clave in d1:
    resultado[clave] = d1[clave] * d2[clave]

print(resultado)
```

```python
# 2) Multiplicar todos los valores por un escalar
diccionario = {"x": 10, "y": 20, "z": 30}
escalar = 3

resultado = {}
for clave in diccionario:
    resultado[clave] = diccionario[clave] * escalar

print(resultado)
```

```python
# 3) Producto punto entre dos vectores representados como diccionarios
v1 = {"a": 1, "b": 2, "c": 3}
v2 = {"a": 4, "b": 5, "c": 6}

producto_punto = 0
for clave in v1:
    print(v1[clave] * v2[clave])
    producto_punto += v1[clave] * v2[clave]

print("Producto punto:", producto_punto)
```

```python
# 4) Costo total por producto y costo general
precios = {"manzana": 1000, "pera": 1500, "uva": 2000}
cantidades = {"manzana": 3, "pera": 2, "uva": 5}

costo_por_producto = {}
costo_total = 0

for producto in precios:
    costo = precios[producto] * cantidades[producto]
    costo_por_producto[producto] = costo
    costo_total += costo

print("Costo por producto:", costo_por_producto)
print("Costo total general:", costo_total)
```
