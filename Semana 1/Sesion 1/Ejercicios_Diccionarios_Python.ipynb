# Ejercicios de Diccionarios en Python



```python
# 1) Promedio de calificaciones
estudiantes = {
    "Ana": 4.5,
    "Luis": 3.8,
    "Sofía": 4.2,
    "Carlos": 3.5,
    "Mateo": 4.9
}

suma = sum(estudiantes.values())
promedio = suma / len(estudiantes)
print(f"Promedio: {promedio:.2f}")
```

```python
# 2) Incrementar precios 10%
productos = {"pan": 2000, "leche": 3800, "huevos": 12000}

for producto in productos:
    productos[producto] *= 1.10

print(productos)
```

```python
# 3) Clave con valor máximo
datos = {"a": 10, "b": 35, "c": 22, "d": 35}

clave_max = max(datos, key=datos.get)
print(f"Clave con valor máximo: {clave_max} (valor={datos[clave_max]})")
```

```python
# 4) Invertir diccionario
original = {"colombia": "CO", "mexico": "MX", "peru": "PE"}

invertido = {valor: clave for clave, valor in original.items()}
print(invertido)
```

```python
# 5) Eliminar palabras con frecuencia < 3
frecuencias = {"hola": 5, "que": 2, "tal": 3, "bro": 1, "python": 4}

frecuencias_filtrado = {k: v for k, v in frecuencias.items() if v >= 3}
print(frecuencias_filtrado)
```

```python
# 6) Combinar diccionarios sumando claves comunes
d1 = {"a": 10, "b": 20, "c": 30}
d2 = {"b": 3, "c": 7, "d": 100}

combinado = d1.copy()

for clave, valor in d2.items():
    combinado[clave] = combinado.get(clave, 0) + valor

print(combinado)
```

```python
# 7) Ordenar diccionario por valores
datos = {"a": 10, "b": 3, "c": 7, "d": 1}

ordenado = dict(sorted(datos.items(), key=lambda item: item[1]))
print(ordenado)
```

```python
# 8) Verificar si una clave existe
capitales = {"Colombia": "Bogotá", "México": "CDMX", "Perú": "Lima"}

clave_usuario = input("Ingresa un país: ")

if clave_usuario in capitales:
    print(f"Sí existe. Capital: {capitales[clave_usuario]}")
else:
    print("No existe esa clave en el diccionario.")
```
